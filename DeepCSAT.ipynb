## 3. ***Data Wrangling***

####Handling Missing Values

````python
# Drop only truly useless columns
df.drop(columns=['Unique id','Order_id'], inplace=True, errors='ignore')

# Categorical
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Numerical
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
````

````python
df.isnull().sum()
````

Missing values were handled using mode and median. Initial implementation raised chained assignment warnings, which were resolved by using direct assignment to ensure clean and future-compatible code.

Instead of dropping all columns with missing values, only irrelevant columns such as IDs were removed. Important features were retained and missing values were handled using imputation techniques to preserve useful information.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Hypothesis testing was performed to statistically validate the relationships observed during exploratory data analysis and to ensure that the findings are not due to random variation.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H0: Response time doesn’t affect CSAT

H1: Response time affects CSAT

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr

temp = df[['response_time', 'csat_score']].dropna()

corr, p_value = pearsonr(temp['response_time'], temp['csat_score'])

print("Correlation:", corr)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0 → Response time affects CSAT")
else:
    print("Fail to reject H0")
````

Since the p-value is very small (< 0.05), we reject the null hypothesis. This means response time affects CSAT. The correlation is negative but weak, so higher response time slightly reduces satisfaction.

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation

##### Why did you choose the specific statistical test?

The Pearson correlation test was chosen because both response time and CSAT score are numerical variables. This test helps measure the strength and direction of the linear relationship between the two variables.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H0: Category has no impact on CSAT

H1: Category affects CSAT

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import f_oneway

groups = [group['csat_score'].values for name, group in df.groupby('category')]

f_stat, p_value = f_oneway(*groups)

print("F-stat:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0 → Category affects CSAT")
else:
    print("Fail to reject H0")
````

The p-value is very small (< 0.05), so we reject the null hypothesis. This means category has a significant impact on CSAT. Different types of issues lead to different satisfaction levels.

##### Which statistical test have you done to obtain P-Value?

ANOVA (Analysis of Variance)

##### Why did you choose the specific statistical test?

ANOVA was selected because the independent variable (category) is categorical with multiple groups, while the dependent variable (CSAT score) is numerical. ANOVA helps determine whether there are statistically significant differences between the means of multiple groups.

### 1. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Capping using IQR
for col in ['item_price','connected_handling_time','response_time']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    df[col] = df[col].clip(lower, upper)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Outliers were handled using the IQR capping method. This prevents extreme values from affecting the model while retaining most of the data.

## ***7. ML Model Implementation***

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate(y_test, y_pred):
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print("MAE:", mae)
    print("MSE:", mse)
    print("RMSE:", rmse)
    print("R2 Score:", r2)
    return mae, rmse, r2
````
